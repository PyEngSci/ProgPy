What Python is seeing is this:  
On the line below the word print is set to the value 42 (an integer)
```python
print = 42
```
After this we are trying to use whatever is represented by the word `print` as a function
```python
print("Hello World!")
```

As `print` is now **not a function** but the *integer value* `42` Python sees this as:
```python
42("Hello World!")
```

Which does not make any sense in Python!
