## Why does `print` fail?

The problem is that we've **overwritten the built-in `print` function**:

```python
print = 42
```

After this line, the name `print` no longer refers to Python's built-in printing function. It now refers to the integer `42`.

So when Python reaches:

```python
print("Hello World!")
```

it effectively tries to do:

```python
42("Hello World!")
```

However, integers can not be used (or *call*ed into use) as functions, so Python raises the error:

```text
TypeError: 'int' object is not callable
```

### Example

```python
print = 42
print("Hello World!")
```

Output:

```text
TypeError: 'int' object is not callable
```

### How to fix it

Use a different variable name:

```python
value = 42
print("Hello World!")
```

Or restore the built-in function:

```python
del print
print("Hello World!")
```

After `del print`, Python falls back to the built-in `print()` function again.

### Key takeaway

Avoid using the names of built-in Python functions as variable names:

```python
list = [1, 2, 3]     # bad
str = "hello"        # bad
sum = 10             # bad
print = 42           # bad
```

Otherwise, you lose access to the original function until the name is deleted or the interpreter is restarted.
